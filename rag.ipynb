```python
import os
from dotenv import load_dotenv
from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.embeddings import JinaEmbeddings
from langchain_community.vectorstores import FAISS
from langchain_groq import ChatGroq
```

```python
load_dotenv()
```

```python
groq_key = os.getenv('GROQ_API_KEY')
jina_key = os.getenv('JINA_API_KEY')
```

```python
#load data
HR_POCILY_DATAFILE = os.path.join('data','hr_policy.txt')
```

#Data ingestion

```python
text_loader = TextLoader(HR_POCILY_DATAFILE,encoding="utf-8")

documents = text_loader.load()
```

#splitting


```python
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size= 100,
    chunk_overlap = 50
)
```

```python
chunks = text_splitter.split_documents(documents)
```

```python
print(chunks)
len(chunks)
```

#Embedding

```python
embedding_model = JinaEmbeddings(model_name="jina-embeddings-v2-base-en",jina_api_key=jina_key)
print("embedding model",embedding_model.model_name)
```

#vector store 

```python
vector_store = FAISS.from_documents(documents=documents,embedding=embedding_model)
print(vector_store.index.ntotal)
```

```python

test_query = "policy"

result = vector_store.similarity_search(query=test_query,k=3)

print(result)
for i,match in enumerate(result,start=1):
    print(f"-----macth{i} ---")
    print(match.page_content)
    print()
```
